```sql
%python

from pyspark.sql.functions import col

posts_por_usuario = spark.read.parquet(
    "dbfs:/Volumes/workspace/default/volume_silver/posts"
)

posts_por_usuario.printSchema()

posts_por_usuario = posts_por_usuario.groupBy("user_id").count()

posts_por_usuario = posts_por_usuario.withColumnRenamed(
    "count", "total_posts"
)

display(posts_por_usuario)
```

```sql
%python

comentarios_por_post = spark.read.parquet(
    "dbfs:/Volumes/workspace/default/volume_silver/comments"
)

comentarios_por_post.printSchema()

comentarios_por_post = comentarios_por_post.groupBy("post_id").count()

comentarios_por_post = comentarios_por_post.withColumnRenamed(
    "count", "total_comentarios"
)

display(comentarios_por_post)
```

```sql
%python

spark.sql("""
CREATE EXTERNAL VOLUME IF NOT EXISTS workspace.default.volume_gold
LOCATION 's3://breno-projeto-dados-2026/gold/'
""")

caminho_gold = "/Volumes/workspace/default/volume_gold/"

posts_por_usuario = posts_por_usuario.write.mode("overwrite").parquet(caminho_gold + "posts_por_usuario")

comentarios_por_post = comentarios_por_post.write.mode("overwrite").parquet(caminho_gold + "comentarios_por_post")

print("Camada Gold gravada com sucesso!")

display(dbutils.fs.ls(caminho_gold))
```
