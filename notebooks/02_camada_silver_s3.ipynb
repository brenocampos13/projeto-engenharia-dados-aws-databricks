```sql
%python
from pyspark.sql.functions import col, trim, lower

df_users = spark.read.json(
    "dbfs:/Volumes/workspace/default/volume_bronze/users.json"
)

df_users_silver = df_users.select(
    col("id"),
    trim(col("name")).alias("name"),
    trim(col("username")).alias("username"),
    lower(trim(col("email"))).alias("email"),
    trim(col("phone")).alias("phone"),
    trim(col("website")).alias("website"),
    trim(col("address.street")).alias("address_street"),
    trim(col("address.suite")).alias("address_suite"),
    trim(col("address.city")).alias("address_city"),
    trim(col("address.zipcode")).alias("address_zipcode"),
    col("address.geo.lat").cast("double").alias("latitude"),
    col("address.geo.lng").cast("double").alias("longitude"),
    trim(col("company.name")).alias("company_name"),
    trim(col("company.catchPhrase")).alias("company_catchPhrase"),
    trim(col("company.bs")).alias("company_bs")
)

display(df_users_silver)
```

```sql
%python
from pyspark.sql.functions import col, count, countDistinct

df_users_silver.select(
    count("*").alias("total_registros"),
    count("id").alias("ids_preenchidos"),
    countDistinct("id").alias("ids_unicos")
).show()
```

```sql
%python
from pyspark.sql.functions import col, trim, lower

df_posts_silver = spark.read.json(
    "dbfs:/Volumes/workspace/default/volume_bronze/posts.json"
)

display(df_posts_silver)

df_posts_silver = df_posts_silver.select(
    trim(lower(col("body"))).alias("body"),
    col("id").alias("id"),
    trim(lower(col("title"))).alias("title"),
    col("userId").alias("user_id")
)

display(df_posts_silver)
```

```sql
%python
from pyspark.sql.functions import col, trim, lower

df_comments_silver = spark.read.json(
    "dbfs:/Volumes/workspace/default/volume_bronze/comments.json"
)

df_comments_silver = df_comments_silver.select(
    trim(lower(col("body"))).alias("body"),
    trim(lower(col("email"))).alias("email"),
    col("id").alias("id"),
    trim(lower(col("name"))).alias("name"),
    col("postId").alias("post_id")
)

display(df_comments_silver)
```

```sql
%python
df_posts_silver.printSchema()
df_comments_silver.printSchema()

print(f"Posts: {df_posts_silver.count()}")
print(f"Comments: {df_comments_silver.count()}")
```

```sql
%python

from pyspark.sql.functions import col, count, countDistinct

def validar_id(df, nome):
    print(f"\nValidando {nome}")
    
    df.select(
        count("*").alias("total_registros"),
        count("id").alias("ids_preenchidos"),
        countDistinct("id").alias("ids_unicos")
    ).show()

validar_id(df_users_silver, "USERS")
validar_id(df_posts_silver, "POSTS")
validar_id(df_comments_silver, "COMMENTS")
```

```sql
%python
posts_sem_usuario = df_posts_silver.join(
    df_users_silver,
    df_posts_silver.user_id == df_users_silver.id,
    "left_anti"
)

comments_sem_post = df_comments_silver.join(
    df_posts_silver,
    df_comments_silver.post_id == df_posts_silver.id,
    "left_anti"
)

print(f"Posts sem usuário: {posts_sem_usuario.count()}")
print(f"Comments sem post: {comments_sem_post.count()}")
```

```sql
%python
spark.sql("""
CREATE EXTERNAL VOLUME IF NOT EXISTS workspace.default.volume_silver
LOCATION 's3://breno-projeto-dados-2026/silver/'
""")
```

```sql
%python

caminho_silver = "/Volumes/workspace/default/volume_silver/"

df_users_silver.write.mode("overwrite").parquet(caminho_silver + "users")
df_posts_silver.write.mode("overwrite").parquet(caminho_silver + "posts")
df_comments_silver.write.mode("overwrite").parquet(caminho_silver + "comments")

print("Camada Silver gravada com sucesso!")

display(dbutils.fs.ls(caminho_silver))
```
